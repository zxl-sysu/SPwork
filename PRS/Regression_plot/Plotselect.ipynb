```python
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('seaborn-v0_8-whitegrid')
nrows=2
ncols=2
main_trait='FG'
fig, axes = plt.subplots(nrows, ncols, figsize=(5*ncols, 4*nrows), squeeze=False, dpi=600,sharey='all')
#fig.suptitle(f'Linear Regression Beta for PRS Score\nTrait: {main_trait}', fontsize=16)
ax_list=[j for i in axes for j in i]
for n,ax in enumerate(ax_list):
    trait=list(df_continue.keys())[n]
    ax.set_title(trait,fontsize=10)
    df_all_param=df_continue[trait].copy()
    df_all_param=df_all_param.sort_values(by=['p_thresh','r2'])
    c=0
    colors=['rosybrown','burlywood','darkseagreen','lightblue']
    n=0
    for p,sub_df in df_all_param.groupby('p_thresh'):
        
        #display(sub_df)
        p_labels = [('%.0e' % p).replace('e', "E") for p in sub_df['p_thresh']]
        c+=0.2
        x = np.arange(len(sub_df))+c
        #print(x)
        neg_log_p = -np.log10(sub_df['p_value'])
        bars = ax.bar(x, neg_log_p, edgecolor='black',width=0.2,label=p,color=colors[n])
        n+=1
        ax.set_xticks(x)
        ax.legend(title='P threshold',fontsize=5,title_fontsize=5)
    x = np.arange(len(sub_df))+0.5
    ax.set_xticks(x)
    r2_labels=sub_df['r2'].apply(lambda x: str(x) ).tolist()
    ax.set_xticklabels(r2_labels)
    ax.set_xlabel('r$^{2}$ threshold')
    ax.set_ylabel('-log$ _{10}$ (P value)')
    ax.hlines(-np.log10(0.05),linestyles='dashed',color='dimgray',xmin=0,xmax=4.5,linewidth=0.8)
    ax.text(4,1.39,'P<0.05',fontsize=10)
    
    #break
    
plt.subplots_adjust(wspace=0.08, hspace=0.2)
plt.savefig(r'D:\Documents\yuan\SPT2D文章修改1103\SPt2d\SPt2d\ValidationPlot\Fig5suppPRS\SELECT_PRS.jpg')
plt.savefig('SELECT_PRS.jpg')
plt.show()
```
